```python
import pandas as pd
import numpy as np

students = {
    "Student_ID": list(range(101, 121)),
    "Name": [
        "Ayesha", "Rahul", "Sara", "Arjun", "Zoya",
        "Rohan", "Priya", "Ali", "Meera", "Kiran",
        "Fatima", "Vikram", "Ananya", "Sameer", "Isha",
        "Kabir", "Neha", "Imran", "Diya", "Adil"
    ],
    "Department": ["CSE-DS"] * 20,
    "Python": [92, 65, 78, 45, 88, 55, 95, 35, 72, 81,
               67, 49, 90, 58, 76, 84, 62, 40, 97, 70],
    "Mathematics": [90, 70, 75, 50, 85, 60, 98, 30, 68, 79,
                    72, 45, 93, 55, 80, 88, 65, 42, 96, 74],
    "Statistics": [88, 68, 80, 42, 90, 58, 92, 38, 75, 83,
                   64, 48, 95, 60, 78, 86, 66, 44, 94, 72],
    "Attendance": [95, 80, 85, 65, 92, 75, 98, 55, 82, 90,
                   78, 60, 96, 72, 88, 91, 77, 58, 99, 84]
}

df = pd.DataFrame(students)

df.to_csv("students.csv", index=False)

print("Dataset created successfully!")
display(df)
```

```python
# Calculate total and average marks

subjects = ["Python", "Mathematics", "Statistics"]

df["Total_Marks"] = df[subjects].sum(axis=1)
df["Average_Marks"] = df[subjects].mean(axis=1)

display(df[["Name", "Total_Marks", "Average_Marks"]])
```

```python
# Function to assign grades

def calculate_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"

# Assign grades to students
df["Grade"] = df["Average_Marks"].apply(calculate_grade)

# Assign pass or fail status
df["Result"] = np.where(df["Average_Marks"] >= 40, "Pass", "Fail")

# Display results
display(df[["Name", "Average_Marks", "Grade", "Result"]])
```

```python
# Class performance summary

print("===== CLASS PERFORMANCE SUMMARY =====")

print("Total Students:", len(df))
print("Class Average:", round(df["Average_Marks"].mean(), 2))
print("Highest Average:", round(df["Average_Marks"].max(), 2))
print("Lowest Average:", round(df["Average_Marks"].min(), 2))

passed = (df["Result"] == "Pass").sum()
failed = (df["Result"] == "Fail").sum()

print("Students Passed:", passed)
print("Students Failed:", failed)
```

```python
# Find the top 5 students

top_students = df.sort_values(
    by="Average_Marks", ascending=False
).head(5)

print("===== TOP 5 STUDENTS =====")
display(top_students[["Name", "Average_Marks", "Grade"]])
```

```python
# Subject-wise performance analysis

subjects = ["Python", "Mathematics", "Statistics"]

print("===== SUBJECT-WISE PERFORMANCE =====")

for subject in subjects:
    average = np.mean(df[subject])
    highest = np.max(df[subject])
    lowest = np.min(df[subject])

    print("\nSubject:", subject)
    print("Class Average:", round(average, 2))
    print("Highest Mark:", highest)
    print("Lowest Mark:", lowest)

best_subject = df[subjects].mean().idxmax()

print("\nBest Performing Subject:", best_subject)
```

```python
# Save complete student analysis

df.to_csv("student_analysis_results.csv", index=False)

print("Analysis results saved successfully!")
print("File name: student_analysis_results.csv")
```
